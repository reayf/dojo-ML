## Langkah 6 — Target (`species`)

**Latihan P.3** — Proporsi tiap spesies → `proporsi_spesies`.

```python
proporsi_spesies = df['species'].value_counts(normalize=1)
cek('P.3', proporsi_spesies)
proporsi_spesies
```

```python
# countplot species di sini
sns.countplot(data=df, x='species')
```

## Langkah 9 — Hubungan fitur ↔ target

**Latihan P.4** — Rata-rata `body_mass_g` per `species` → `massa_per_spesies`.

```python
hint('P.6')
```

```python
massa_per_spesies = df.groupby('species')['body_mass_g'].mean()
cek('P.4', massa_per_spesies)
massa_per_spesies
```

**Latihan P.6** — Tabel korelasi antar kolom angka → `kor`.

```python
kor = df.corr(numeric_only=True)
cek('P.6', kor)
kor
```

```python
# boxplot flipper_length_mm per species, crosstab island vs species, dll.

sns.boxplot(data = df, x='species',y = 'flipper_length_mm')
plt.title('Distribusi Flipper Length per Spesies')
plt.xlabel('Spesies')
plt.ylabel('Flipper Length (mm)')
plt.show()

```
